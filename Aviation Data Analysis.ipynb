### Does Price Varies With Airline?

```python
data.head()
```

```python
#Grouping the airline to check their mean price

data.groupby('airline')['price'].mean()
```

```python
#Drawing a categorical plot showing the mean ticket price for each airline and the classes

sns.catplot( x = 'airline', y = 'price', kind = 'bar', data = data, hue = 'class')
plt.show()
```

### Does Departure and Arrival Time affect Ticket Price?

```python
data.head()
```

```python
#Checking the mean ticket price based on departure time

data.groupby('departure_time')['price'].mean()
```

```python
#Checking the mean ticket price based on arrival time

data.groupby('arrival_time')['price'].mean()
```

```python
sns.catplot( x = 'departure_time', y = 'price', kind = 'bar', data = data)
plt.show()
```

```python
sns.catplot( x = 'arrival_time', y = 'price', kind = 'bar', data = data)
plt.show()
```

```python
#Show the Departure time and arrival time in a relational plot

sns.relplot(x = 'arrival_time', y='price', data = data, col = 'departure_time')
plt.show()
```

```python
#Showing the relational price of Departure time and arrival time in a line plot

sns.relplot(x = 'arrival_time', y='price', data = data, col = 'departure_time', kind = 'line')
plt.show()
```
